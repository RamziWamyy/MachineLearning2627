````python
# Accessing a file using file locks to get variable names
with open(osp.join('..','data','mia','miaPassengers.names'),'r', encoding='utf-8-sig') as f:
    columns = f.read().strip().split(',')

columns
````

We created a pointer ```f``` to the file we wnat to access and attached a read lock. We also define the encoding, meaning the character space to use to interpret the context, utf8. the ```utf-8-sig``` variant is needed to support the specific BOM character of this file. Every file has a BOM and EOF character.

When reading the context of the file we remove whitespaces using the ```strip()``` method, then we create a list with the contents of the file using the ```split(',')``` mehtod since they are comma sepearted.

````python
df = pd.read_csv(osp.join('..','data','mia','miaPassengers.data'),names=columns)
df.info()
df.head()
````
